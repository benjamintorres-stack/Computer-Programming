### when i use a for, the word "for" is a moving variable. It will get set to a new item every time the loop runs. This word does not exist outside the loop.

```python
perfect_squares = []
for i in range(1,11):
    perfect_squares.append(i**2)
print(perfect_squares)
```

### When i use a for loop, the i changes everytime the loops runs. i can use the i somewhere else if necessary.

```python
fish = ("bass","grouper","salmon","yellowtail","tuna","trout")
print(fish[1:4])
```
