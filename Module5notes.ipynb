```python
answer = 3
year = int(input("Pick a number 1-10"))
count = 1
while year != answer:
    print("wrong, try again")
    count = count + 1
    year = int(input("guess again"))
print("You won")
if count == 1:
    print("It took you 1 try")
else:
    print("It took you",count,"tries")
```
