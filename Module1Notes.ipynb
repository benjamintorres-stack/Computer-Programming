### Pop requires the index of the value that you remove. Remove needs the exact value.

```python
activity = ["one","two","three","four","five"]
activity.append("hello")
print(activity)
activity.insert(3,1)
print(activity)
del(activity[1])
print(activity)
del(activity[-1])
print(activity)
y = activity.pop()
print(f"the last word in my list was {y}")
```

### Pop has 3 funtions. 1 it can remove by index, ex:activity.pop(1). It can store a value it deletes for a second so you can store it in a variable. EG y = activity.pop(-1). 3 it can make the executive decision to delete the last item if you don't assign a value. EG my_list.pop()
